```python
target_dir = r"C:\\1.My Files\\Paing Lin Htike\\School\\Semester-8\\CS - 601\\Nutrify\\data/100_foods"
```

```python
import os
import pathlib # see here for more on PathLib - https://realpython.com/python-pathlib/
filepaths = []
for path, subdirs, files in os.walk(target_dir):
    for name in files:
        filepaths.append(pathlib.Path(path, name))
```

```python
len(filepaths)
```

```python
import random
import numpy as np
np.random.seed(42)
random_indexs = list(range(len(filepaths)))
np.random.shuffle(random_indexs)
random_indexs[:10]
```

```python
train_split = round(0.8 * len(filepaths))
train_idx = random_indexs[:train_split]
test_idx = random_indexs[train_split:]

train_files = np.array(filepaths)[train_idx]
test_files = np.array(filepaths)[test_idx]

len(train_files) + len(test_files)
```

```python
os.makedirs(r"C:\\1.My Files\\Paing Lin Htike\\School\\Semester-8\\CS - 601\\Nutrify\\100_whole_foods/train", exist_ok=True)
os.makedirs(r"C:\\1.My Files\\Paing Lin Htike\\School\\Semester-8\\CS - 601\\Nutrify\\100_whole_foods/test", exist_ok=True)
```

```python
class_list = sorted(list(set([filepath.parent.stem for filepath in filepaths])))
class_list
```

```python
# Make directory in train and test folder
train_dir = pathlib.Path(r"C:\\1.My Files\\Paing Lin Htike\\School\\Semester-8\\CS - 601\\Nutrify\\100_whole_foods/train/")
test_dir = pathlib.Path(r"C:\\1.My Files\\Paing Lin Htike\\School\\Semester-8\\CS - 601\\Nutrify\\100_whole_foods/test/")
split_dirs = [train_dir, test_dir]

for targ_dir in split_dirs:
    for targ_class in class_list:
        os.makedirs(pathlib.Path(targ_dir, targ_class)) 
```

```python
from PIL import Image
Image.MAX_IMAGE_PIXELS = None  # Allow large Pexels images

SOURCE = r"C:\1.My Files\Paing Lin Htike\School\Semester-8\CS - 601\Nutrify\data\100_foods"
TARGET = (380, 380)

for root, _, files in os.walk(SOURCE):
    for name in files:
        if name.lower().endswith((".jpg", ".jpeg", ".png")):
            path = os.path.join(root, name)
            try:
                img = Image.open(path)
                img = img.resize(TARGET, Image.LANCZOS)
                img.save(path, quality=85, optimize=True)
            except:
                pass
```

```python
from shutil import copy2
import pathlib

def copy_to_dir(targ_files, targ_dir):
    """Copies targ_files to targ_dir"""
    # Copy train files to train dir
    for filepath in list(targ_files):
        targ_class = filepath.parent.stem
        targ_file = filepath.name
        targ_dir_class = pathlib.Path(targ_dir, targ_class)
        # Copy target file to target dir
        print(f"Copying: {filepath} to: ... ")
        destination = pathlib.Path(targ_dir_class, targ_file)
        print(destination)
        copy2(filepath, destination)
```

```python
copy_to_dir(targ_files=train_files, targ_dir=train_dir)
```

```python
copy_to_dir(targ_files=test_files, targ_dir=test_dir)
```

```python
print(f"Train: {len(train_files)} ({len(train_files)/len(filepaths)*100:.1f}%)")
print(f"Test:  {len(test_files)} ({len(test_files)/len(filepaths)*100:.1f}%)")
print(f"Total: {len(train_files) + len(test_files)}")
```
